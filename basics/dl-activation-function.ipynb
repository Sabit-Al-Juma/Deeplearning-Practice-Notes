# Sigmoid

**If we use any value using signmoid function, itll give output between 0 and 1**

```python
def sigmoid(X):
    return 1 / (1 + math.exp( -X ))
```

```python
sigmoid(1)
```

```python
sigmoid (-45)
# e-20, which is very very close to zero
```

```python
sigmoid (0.1)
```

# tanh

**tanh is better than sigmoid function. All other places try to use sigmoid function. it gives you the output between -1 to 1.**

```python
def tanh(x):
    return (math.exp(x) - math.exp(-x)) / (math.exp(x) + math.exp( -x))
```

```python
tanh(-34)
```

```python
tanh(1)
```

```python
tanh(0)
```

# Relu

**Relu is very light weight function. But it has a condition. if relu(x) <= 0, itll give the output 0, else itll send same as input.**

```python
def relu(x):
    return max(0, x)
```

```python
relu(1)
```

```python
relu(43)
```

```python
relu(0)
```

# Leaky Relu

```python
relu(-34)
```

```python
def leaky_relu(x):
    return max((0.1*x), x)
```

```python
leaky_relu(1)
```

```python
leaky_relu(-34)
```

```python
leaky_relu(0)
```

```python
leaky_relu(-1)
```
